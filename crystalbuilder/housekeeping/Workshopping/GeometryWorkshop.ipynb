```python
a1 = [1, 1, 0]
a2 = [0, 1 ,1]
a3 = [1, 0, 1]

a_mag = 1

geo_lattice = lattice.Lattice(a1, a2, a3, magnitude = [a_mag, a_mag, a_mag])
```

```python
vec1 = [1,1,0]
vec2 = [0,1,0]
vec3 = [0,0,1]
size = .25
block = geo.Block(center=[0,0,0], vectors = [vec1, vec2, vec3], size=[size,size, size])
block2 = geo.Block(center=[.5,0,0], vectors = [vec1, vec2, vec3], size=size)
block3 = geo.Block(center=[0,0,.5], vectors = [vec1, vec2, vec3], size=[.3, .3, size])

unit_cell=[block, block2, block3]
```

```python
a1_reps = 1
a2_reps = 1
a3_reps = 1
crystal = geo_lattice.tile_geogeometry(unit_cell, a1_reps, a2_reps, a3_reps )
```

```python
scene = viewer.visualize(crystal)
scene.show().close()
```
