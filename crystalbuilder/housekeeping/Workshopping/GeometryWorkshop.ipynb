```python
a1 = [1, 0, 0]
a2 = [0, 1 ,0]
a3 = [0, 0, 1]

a_mag = 1

geo_lattice = lattice.Lattice(a1, a2, a3, magnitude = [a_mag, a_mag, a_mag])
```

```python
block = geo.Block(center=[0,0,0], vectors = [a1, a2, a3], size=[.1,.1,.1])
unit_cell=[block]
```

```python
a1_reps = 2
a2_reps = 2
a3_reps = 2
crystal = geo_lattice.tile_geogeometry(unit_cell, a1_reps, a2_reps, a3_reps )
```

```python
scene = viewer.visualize(crystal)
scene.show().close()
```
