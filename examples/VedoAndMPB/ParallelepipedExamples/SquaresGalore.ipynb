## Example 1: A Cubic Lattice of Cubes <a id="example1"></a>

Here is a simple cubic array with cubic atoms. As always, we start with our lattice vectors. Here they're just the Cartesian bassis vectors (x, y, and z)

```python
a1 = [1, 0, 0]
a2 = [0, 1 ,0]
a3 = [0, 0, 1]

a_magnitude = 1


geo_lattice = lattice.Lattice(a1, a2, a3, magnitude = [a_magnitude, a_magnitude, a_magnitude])
```

We define our cubes using the geo.Block() function. The function takes the center position of the cube and a set of vectors that define the direction of the edges. This matches the [MEEP/MPB Convention](https://meep.readthedocs.io/en/master/Python_User_Interface/#block). 

An important difference, however, is that you can also have your size defined by these vectors. MEEP/MPB automatically discards the magnitudes of the vectors, but CrystalBuilder uses their magnitudes to define the size by default. If you want to override this, simply pass a size argument to the function. See [the appendix](#resized-cube) for examples of the syntax.

Here we do want to override this, since a 1x1x1 cube would fill the unit cell!

```python
cube_vectors = [
    [1,0,0],
    [0,1,0],
    [0,0,1]
]

cube = geo.Block(center=(0,0,0), vectors=cube_vectors, size=.25)
```

Before we simulate this with MEEP/MPB, let's look at it in a viewing window. We will use Vedo for this example, but Trimesh is another option (and likely works better for outputs)

MPB takes a single unit cell and builds the lattice infinitely. For our viewer, we should specify how many unit cells we want to look at.

```python
a1_reps = 2
a2_reps = 2
a3_reps = 2
cube_crystal = geo_lattice.tile_geogeometry(cube, a1_reps, a2_reps, a3_reps )
```

```python
scene = viewer.visualize(cube_crystal)
scene.show()
```

## Example 2: A Cubic Lattice of Rhombohedrons <a id="example2"></a>

This is very similar to [Example 1](#example1), but demonstrates how the lattice vectors and parallelepiped vectors can be different

```python
a1 = [1, 0, 0]
a2 = [0, 1 ,0]
a3 = [0, 0, 1]
a_magnitude = 1
geo_lattice = lattice.Lattice(a1, a2, a3, magnitude = [a_magnitude, a_magnitude, a_magnitude])
```

The difference in this example is the vector set that defines our parallelepiped. 
We'll use:
1. <1,1,0>
2. <0,1,1>
3. <1,0,1>

```python
rhombohedron_vectors = [
    [1,1,0],
    [0,1,1],
    [1,0,1]
]

rhomb = geo.Block(center=(0,0,0), vectors=rhombohedron_vectors, size=.25)
```

```python
a1_reps = 2
a2_reps = 2
a3_reps = 2
rhombohedron_crystal = geo_lattice.tile_geogeometry(rhomb, a1_reps, a2_reps, a3_reps )
```

```python
scene = viewer.visualize(rhombohedron_crystal)
scene.show()
```

## Appendix 

##### Resized Cube

1. A "cube" with no specified size. It will be scaled to whatever the magnitudes of cube_vectors are

        Cube = geo.Block(center=(0,0,0), vectors=cube_vectors)

2. A "cube" with 3 specified sizes which scales the respective 3 vectors. If they are different sizes, it's no longer a cube!

        Cube = geo.Block(center=(0,0,0), vectors=cube_vectors, size=(1,1,1))

3. A cube with a single specified size, scaling all vectors by this value

        Cube = geo.Block(center=(0,0,0), vectors=cube_vectors, size=2)
